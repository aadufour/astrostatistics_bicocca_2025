##compute the area of a N-dimensional sphere

```python
# Execute this cell
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt
from tqdm import tqdm
%matplotlib inline
%config InlineBackend.figure_format='retina'
```

```python
if False:

    NDIM = 10
    M = 10000

    #area of "sphere"/area of "square"
    ratio = []
    for dim in tqdm(range(NDIM)):
        
        #initialising array and calculating norms
        x = np.random.uniform(0, 1, (dim+1, M))
        normx = np.linalg.norm(x, axis=0)

        ratio.append(np.sum([normx<1])/M)

    plt.scatter(np.arange(NDIM)+1, ratio)
    # plt.yscale("log")


```

```python
def predict(current):
    chance = np.random.uniform(0, 1)
    if current == 1: #today is sunny
        if chance<=0.9: return 1
        else: return 0
    if current == 0:
        if chance <= 0.5: return 1
        else: return 0
    else: print("current must be 0 or 1")

N=50000
current = 1

log = []
trace = []
# trace.append(current)

for _ in tqdm(range(N)):
    current = predict(current)
    log.append(current)
    trace.append(np.sum(log)/(len(log)))


plt.plot(range(N), trace, label="no burn")

#------------ burn ------------
burn = 3000
plt.plot(range(burn, N), trace[burn:], label=f"burn={burn}")


plt.xlabel("days")
plt.ylabel(" p(clear) [% of sunny days]")
plt.legend()
plt.show()
```

```python
#----------- histograms -------------
plt.hist(trace, label="no burn", bins="scott")
plt.hist(trace[burn:], label=f"burn={burn}", bins="scott")

# plt.xlim(0.82, 0.84)

plt.xlabel("p(clear)")
plt.show()
```
